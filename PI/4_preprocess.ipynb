```python
import pandas as pd
df = pd.read_csv('3_output_manual_labeled_user_comments_rev.csv')
df.head()
```

```python
import json

# Muat slang words ke dalam dictionary
slang_dict = {}
with open('combined_slang_words.txt', 'r') as f:
    for line in f:
        parts = line.strip().split('\t')
        content = parts[0]
        slang_dict = json.loads(content)
print(slang_dict)
        
```

```python
len(slang_dict)
```

```python
import nltk
# --- Setup ---
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
nltk.download('punkt_tab')
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

stop_words = set(stopwords.words('indonesian'))  # Indo stopwords
lemmatizer = WordNetLemmatizer()
factory = StemmerFactory()
stemmer = factory.create_stemmer()
```

```python
import re
def preprocess_text(text):
    # Case Folding
    text = text.lower()

    # Text Cleansing
    print('text cleansing')
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # Normalisasi (tambahan)
    print('normalisasi')
    tokens = text.split() # Split into tokens first
    normalized_tokens = [slang_dict.get(token, token) for token in tokens]
    text = ' '.join(normalized_tokens)

    # Stop Words
    print('stop words')
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]

    # Stemming
    print('stemming')
    stemmed_text = ' '.join(tokens)
    stemmed_text = stemmer.stem(stemmed_text)
    return stemmed_text
```

```python
# Apply preprocessing
df['clean_text'] = df['comment'].astype(str).apply(preprocess_text)
```
